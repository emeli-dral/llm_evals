# Naive answer tests

Imagine we are building the first version of a support chatbot for a small online shop. We already have a few answers from the chatbot, and we want a very simple smoke test: does each answer include the most important expected words?

This is not a complete evaluation. It is just the smallest possible check: look for substrings in a string.

```python
from pprint import pprint
```

## Tiny substring matcher

We normalize the text so the check ignores letter case and repeated spaces.

```python
def normalize(text):
    return " ".join(text.casefold().split())


def contains_phrase(text, phrase):
    return normalize(phrase) in normalize(text)


def contains_all_phrases(text, phrases):
    return all(map(lambda phrase: contains_phrase(text, phrase), phrases))
```

## The first tests

Each row says: this chatbot answer should include these words or phrases.

```python
test_cases = [
    {
        "answer": "You can return your order through the Returns Center. We will email you a prepaid label.",
        "expected_phrases": ["prepaid label"],
    },
    {
        "answer": "You have 14 days from delivery to return an item.",
        "expected_phrases": ["14 days"],
    },
    {
        "answer": "Yes, we ship worldwide. Delivery times depend on the destination.",
        "expected_phrases": ["worldwide"],
    },
    {
        "answer": "You can contact support at support@example.com.",
        "expected_phrases": ["support@example.com"],
    },
]
```

```python
results = list(map(
    lambda test: {
        "answer": test["answer"],
        "expected_phrases": test["expected_phrases"],
        "passed": contains_all_phrases(test["answer"], test["expected_phrases"]),
    },
    test_cases,
))

pprint(results, sort_dicts=False)
```

```python
passed = sum(map(lambda result: result["passed"], results))
total = len(results)

print(f"Pass rate: {passed / total:.0%}")
```

This kind of test is useful as a first smoke test, but it is brittle. An answer can be correct and still fail if it uses different wording, such as `two weeks` instead of `14 days`.
