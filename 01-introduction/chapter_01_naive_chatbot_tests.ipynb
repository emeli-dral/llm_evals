# Naive scope tests

Imagine we are building the first version of a support chatbot for a small online shop. Before we test the answers themselves, we can test something simpler: do our example questions belong to the topics the chatbot is supposed to handle?

This first test is intentionally naive. It only looks for expected words inside each question.

```python
from pprint import pprint
```

## Topics our chatbot supports

```python
scope_keywords = {
    "returns": ["return", "refund"],
    "shipping": ["ship", "delivery"],
    "support": ["support", "contact"],
    "payments": ["payment", "pay", "card"],
}
```

## Tiny substring matcher

We normalize the text so the check ignores letter case and repeated spaces.

```python
def normalize(text):
    return " ".join(text.casefold().split())


def contains_any(text, keywords):
    text = normalize(text)
    return any(map(lambda keyword: normalize(keyword) in text, keywords))


def matches_scope(question, expected_topic):
    return contains_any(question, scope_keywords[expected_topic])
```

## The first tests

Each row says: this question should be recognized as belonging to this topic.

```python
test_cases = [
    {"question": "How can I return my order?", "expected_topic": "returns"},
    {"question": "When will my package ship?", "expected_topic": "shipping"},
    {"question": "How do I contact support?", "expected_topic": "support"},
    {"question": "Can I pay with a credit card?", "expected_topic": "payments"},
]
```

```python
results = list(map(
    lambda test: {
        "question": test["question"],
        "expected_topic": test["expected_topic"],
        "passed": matches_scope(test["question"], test["expected_topic"]),
    },
    test_cases,
))

pprint(results, sort_dicts=False)
```

```python
passed = sum(map(lambda result: result["passed"], results))
total = len(results)

print(f"Pass rate: {passed / total:.0%}")
```

This is useful as a first smoke test, but it is still brittle. A question like `Can I send this back?` is about returns, but it does not contain `return` or `refund`, so this simple test would miss it.
